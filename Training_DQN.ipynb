```python
import gymnasium as gymn
from stable_baselines3 import DQN
from stable_baselines3.common.monitor import Monitor
import os
import sys

log_dir = "logs/Training_DQN"
os.makedirs(log_dir, exist_ok=True)

base_env = gymn.make("LunarLander-v3")
env = Monitor(base_env, log_dir, allow_early_resets=True)   

model = DQN(
    "MlpPolicy",
    env,
    learning_rate=6.3e-4,
    buffer_size=50000,
    learning_starts=1000,
    batch_size=128,
    gamma=0.99,
    target_update_interval=250,
    train_freq=4,
    gradient_steps=1,
    exploration_fraction=0.12,
    exploration_final_eps=0.1,
    verbose=1,
)

print("Training the model...")
model.learn(total_timesteps=270000)

model.save("dqn_lunarlander")
print("Model saved as dqn_lunarlander.zip")
env.close()
```
