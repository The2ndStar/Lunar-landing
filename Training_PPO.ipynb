```python
import gymnasium as gym
from stable_baselines3 import PPO
from stable_baselines3.common.monitor import Monitor
import os
import sys

log_dir = "./logs/Training_PPO"
os.makedirs(log_dir, exist_ok=True)

env = Monitor(gym.make("LunarLander-v3"), log_dir)

model = PPO(
    "MlpPolicy",
    env,
    learning_rate=3e-4,
    n_steps=1024,
    batch_size=64,
    n_epochs=10,
    gamma=0.999,
    gae_lambda=0.98,
    ent_coef=0.01,
    verbose=1,
)

print("Training the model...")
model.learn(total_timesteps=270000)

# Save explicitly in root directory
model.save("ppo_lunarlander")
print("Model saved as ppo_lunarlander.zip")

env.close()
```
